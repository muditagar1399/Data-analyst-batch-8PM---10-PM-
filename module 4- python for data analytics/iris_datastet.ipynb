```python
# using box plot we can find outliers 
```

```python
sns.boxplot(x=data['sepal_length'])
plt.show()
```

```python
q1=data['sepal_length'].quantile(0.25)
q3=data['sepal_length'].quantile(0.75)
iqr=q3-q1
lower_limit=q1-1.5*iqr
upper_limit=q3+1.5*iqr
outliers=data[(data['sepal_length']<lower_limit)|(data['sepal_length']>upper_limit)]
print(outliers)
```

```python
q1=data['sepal_width'].quantile(0.25)
q3=data['sepal_width'].quantile(0.75)
iqr=q3-q1
lower_limit=q1-1.5*iqr
upper_limit=q3+1.5*iqr
outliers=data[(data['sepal_width']<lower_limit)|(data['sepal_width']>upper_limit)]
print(outliers)
```
