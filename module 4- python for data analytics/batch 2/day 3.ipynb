```python
# armstrong numbers
# armstrong numbers also called as narcissistic number is a number that is equal to sum of its own diigts each raise to the power of number of diigits 
```

```python
#eg: 1 power is 3 , 3 power is 3 =1+27=28
```

```python
num=int(input("enter the number : "))
sum=0
for digit in str(num):
    sum +=int(digit)**2
if sum==num:
    print("number is armstrong ")
else :
    print("not armstrong")
    
```

```python
# if i want to check is prime or not 
```

```python
#prime number is a number which has exact two factors for ex: 2, ,5,3,7 all prime prime 
```

```python
num=int(input("enter the number :"))
count =0

for num in range(1, num+1):
    if num%1==0:
        count+=1
if count ==2:
    print("numner is prime")
else:
    print("not prime")
```
