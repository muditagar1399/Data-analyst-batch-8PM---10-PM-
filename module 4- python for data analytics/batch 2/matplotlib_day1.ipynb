# number of mobiles by brand (bar chart)

```python
brand_count=data["Brands"].value_counts()
```

```python
brand_count
```

```python
plt.figure(figsize=(10,6))
brand_count.plot(kind="bar")
plt.title("number of mobiles by brand ")
plt.xlabel("Brand")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

```python
# plt.title = giving the title to the graph
#plt.xlabel - giving the name to xlabel 
#plt.ylabel - to dispaly the y label 
# plt ,xticks is the total number of ticks basucally you can show up to 45 nmaes on xlabel if you are wrtiitng xticks as 45
# plt.show- it is to diaply the graph 
```

#  visual 2 - histogram

```python
plt.figure(figsize=(8,5))
plt.hist(data["Selling Price"],bins=20)
plt.title("Distribution of selling prices ")
plt.xlabel("Selling Price ")
plt.ylabel("Frequency")
plt.show()
```

# pie chart 

```python
plt.figure(figsize=(7,7))
plt.pie(
    brand_count,
    labels=brand_count.index,
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Top 5 brands")
plt.show()
```

```python
# autopct - to show percentage 
# startangle = to show the starting from 90
```

```python
top5=data["Brands"].value_counts().head(5)
```
