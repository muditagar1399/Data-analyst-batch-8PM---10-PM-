```python
#converting the column to numeric 
```

```python
# memory 
```

```python
data["Memory"]=data["Memory"].str.replace("GB", "",regex=False).str.strip()
data["Memory"]=pd.to_numeric(data["Memory"],errors="coerce")
```

```python
data
```

```python
data["Storage"]=data["Storage"].str.replace("GB", "",regex=False).str.strip()
data["Storage"]=pd.to_numeric(data["Storage"],errors="coerce")
```

```python
data
```

```python
data["Memory"]=data["Memory"].fillna(data["Memory"].mean())
```

```python
data
```

```python
data["Storage"]=data["Storage"].fillna(data["Storage"].mean())
```

```python
data["Rating"]=data["Rating"].fillna(data["Rating"].mean())
```
