```python
# to find most selling product based on product id 
```

```python
data=pd.read_csv("ecommerce_10000.csv")
```

```python
data
```

```python
## EDA
```

```python
##finding most selling product based on order id 
```

```python
data.groupby("OrderID")["TotalAmount"].sum()
```

```python
## to check how many products are there per order
```

```python
data.groupby("Product")["OrderID"].count()
```

```python
## to fund the highest selling product based on total amount
```

```python
a=data.groupby("Product")["TotalAmount"].sum()
```

```python
a
```

```python
top_product=a.idxmax()
```

```python
top_product
```

```python
## which brand is most selling and which brand is least selling 
```

```python
brand=data.groupby("Brand")["TotalAmount"].sum().sort_values(ascending=False)
```

```python
brand
```

```python
## sort data in asc and desc order in python 
```
