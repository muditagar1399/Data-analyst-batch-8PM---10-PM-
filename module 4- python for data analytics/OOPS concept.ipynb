```python
#encapulation : encapsulation means hiding data means data at one place
```

```python
class bankaccount:
    def __init__(self,balance):
        self.__balance=balance 
    def deposit(self,amount):
        self.__balance +=amount
    def show_balance(self):
        print("balance:", self.__balance)
acc=bankaccount(1000)
acc.deposit(500)
acc.show_balance()
```

```python
# encapsulation : wrapping data and methods together and restricitng direct access
```

```python
##self.__balance : double underscore __ makes varibale private it cannot be accessed directly outside the class  this protects sensitive data 
```

```python
# inhertiance :
#inherting the priperty of parent into child class 
```

```python
class animal:
    def sound(self):
        print("animal makes sound")
class Dog(animal):
    def bark(self):
        print("dog barks")
d=Dog()
d.sound()
d.bark()
```

```python
# here you inherited class animal with class dog in this case parent class is animal 
```

```python
# polymorphism 
# same method but behaves differently 
```

```python
class cat:
    def sound(self):
        print("cat says meow")
class dog:
    def sound(self):
        print("dog is barking")
for animal in [cat(),dog()]:
    animal.sound()
```

```python
#in this case even the functions name is same but the arguments are different 
```

```python
# abstraction : used to hide implementation details 
```

```python
from abc import ABC,abstractmethod
class vechile(ABC):
    @abstractmethod
    def start(self):
      pass
```

```python
#abc : abstract means class module it is used in python for data abstraction 
```

```python
#ABC : used to create an abstract class 
#abstract method : used to create abstract method 
```

```python
# vechile is the class name it is an abstract class it inherits from ABC 
# it means vechile acts like a blueprint/template for abc 
#you cannot directly create its object 
```

```python
# wherever you use abc method 
class Car(vechile):
    def start(self):
        print("car starts with key")
c=Car()
c.start()
```
