# how to find z score in python 

```python
import pandas as pd 
from scipy.stats import zscore 
```

```python
numeric_cols=['Cost','Sales','Quantity','Total_Cost']
for col in numeric_cols:
    data[f'{col}_zscore']=zscore(data[col])
```
