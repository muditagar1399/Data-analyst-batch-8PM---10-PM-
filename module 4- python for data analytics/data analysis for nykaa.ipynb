# Outliers 

# an outlier is a value in your data that is very different from most of the other values for ex : 20000, 3000,  80000 in this case 800000 is the outlier

# IQR (inter-quartile range) : it is a way to measure the middle spread of your data 
we have 3 key points here :
q1 (25th percentile)- 25% of data below this 
q2=median middle value 
q3=75th percnetile = 75% of data below this 

IQR=Q3-Q1
it tells you spread of the middle 50% of the data 

how iQR HELP IN FINDING OUTLIERS 

we have boundaries here to define = lower limit = q1-1.5*IQR 
UPPER LIMIT : Q3+1.5*IQR 

any value which is outside this range = outlier

it can only be used to check in numerical columns 

```python
Q1=df['Revenue'].quantile(0.25)
Q3=df['Revenue'].quantile(0.75)
IQR=Q3-Q1
outliers =df[(df['Revenue']<Q1-1.5*IQR)|(df['Revenue']+1.5*IQR)]
```
